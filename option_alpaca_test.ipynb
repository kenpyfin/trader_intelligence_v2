```python
import asyncio
from typing import Sequence

from backend.src.my_libs_py3.TD_Order import AlpacaTrader, _extract_expiration_list


```

```python
async def fetch_option_chain(symbol: str) -> Sequence[str]:
    """Return sorted expiration dates (YYYYMMDD) for the provided ticker."""
    trader = AlpacaTrader()
    await trader.connect()
    chain = await trader.client.get_option_chain(symbol)
    expirations = sorted(_extract_expiration_list(chain))
    return expirations
```

```python



async def main() -> None:
    import argparse

    parser = argparse.ArgumentParser(description="Fetch option expirations from Alpaca")
    parser.add_argument("symbol", help="Underlying ticker symbol, e.g. AAPL")
    args = parser.parse_args()

    expirations = await fetch_option_chain(args.symbol)
    if not expirations:
        print(f"No option expirations found for {args.symbol.upper()}.")
        return

    print(f"Option expirations for {args.symbol.upper()}:")
    for expiry in expirations:
        print(f"  {expiry}")
```
